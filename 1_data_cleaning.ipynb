```python
import pandas as pd

# Load raw data
df = pd.read_csv('../data/raw_data.csv')

print("Shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())

print("\nColumn Info:")
print(df.info())
```

```python
# Check for duplicates
print("Duplicate rows:", df.duplicated().sum())

# Check for missing values (already confirmed none, but double-check)
print("\nMissing values:")
print(df.isnull().sum())

# Basic statistics
print("\nStatistical Summary:")
print(df.describe())
```

```python
# Save cleaned data (already clean, no major changes needed)
df.to_csv('../data/cleaned_data.csv', index=False)
print("Cleaned data saved successfully!")
print(f"Final shape: {df.shape}")
```
