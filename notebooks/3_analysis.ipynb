```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv('../data/cleaned_data.csv')

# Summary of all key findings
print("="*50)
print("   CUSTOMER CHURN ANALYSIS - KEY FINDINGS")
print("="*50)
print(f"Total Customers: {len(df)}")
print(f"Overall Churn Rate: {df['Churn'].mean()*100:.2f}%")
print(f"\nChurn by Contract Type:")
print(df.groupby('Contract')['Churn'].mean() * 100)
print(f"\nAvg Tenure - Churned: {df[df['Churn']==1]['Tenure'].mean():.1f} months")
print(f"Avg Tenure - Retained: {df[df['Churn']==0]['Tenure'].mean():.1f} months")
print("="*50)
```
